```python
from gummybear.geometry import inspect_stl

STL = ROOT / "cad" / "proto_bear.stl"
inspection = inspect_stl(STL)
mesh = inspection["mesh"]
summary = inspection["summary"]
validation = inspection["validation"]

summary, validation
```

```python
print("Watertight:", summary["is_watertight"])
print("Winding:", summary["is_winding_consistent"])
print("Area:", summary["area"])
print("Volume:", summary["volume"])
print("Bounds:", summary["bbox_min"], "→", summary["bbox_max"])
print("Extents:", summary["bbox_size"])
print("Centroid:", summary["centroid"])
print("Usable for projection:", validation["usable_for_projection"])
if validation["errors"]:
    print("Errors:", validation["errors"])
if validation["warnings"]:
    print("Warnings:", validation["warnings"])
```
