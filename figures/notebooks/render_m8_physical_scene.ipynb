# M8 physical setup — POV-Ray illustration

Thin notebook: export a POV-Ray 3.7 scene of the **first M8 sample** (bear STL, labelled particle, point light, 180° camera, coordinate plane).

Illustration knobs (yaw, FOV, stand-off, ray count, particle marker size, caption ``(x, y)`` fractions) live in the next cell. After a POV render, re-run the last cell to move captions without ray-tracing again.

Install from the repo root (no editable install): `pip install "." -c requirements.txt`

```python
from pathlib import Path

from pathlib import Path

from gummybear.paths import display_path, repo_relative_path
from gummybear_illustration import export_m8_physical_scene
from gummybear_illustration.caption_overlay import overlay_workflow_captions
from gummybear_illustration.paths import repo_root

ROOT = repo_root()

# Configure the export. sample_index=0 is the first M8 sequence with a manifest.
SAMPLE_INDEX = 0
CAMERA_ANGLE_DEG = 180
OUTPUT_POV = ROOT / "outputs" / "pov" / "m8_physical_scene.pov"
RENDER = True  # True attempts povray if installed; export still works if False

# Illustration POV-Ray camera (not the M8 acquisition pose).
# yaw: 0° = behind the pinhole looking at the bear; increase toward the side overview.
# fov: POV-Ray camera angle in degrees (larger = wider).
# distance_scale: stand-off behind the pinhole, in units of pinhole–bear distance
#   (larger = further back, easier to fit the optical camera in frame).
ILLUSTRATION_YAW_DEG = 80.0
ILLUSTRATION_FOV_DEG = 60.0
ILLUSTRATION_DISTANCE_SCALE = 2.1

# Catalog-light rays (0 = yellow marker only, no cylinders) and particle marker size.
ILLUSTRATION_N_RAYS = 0
ILLUSTRATION_PARTICLE_RADIUS_MM = 2  # None = automatic from mesh; else radius in mm
# Green particle as a local light_source that tints the transmissive bear. 0 = sphere only.
ILLUSTRATION_PARTICLE_LIGHT = 0

# Catalog-light cone: length as a fraction of light-to-bear-centre distance.
# Intensity (emission + opacity) falls to 0 at that length. 0 = no cone.
ILLUSTRATION_LIGHT_CONE_LENGTH_FRAC = 0.2

# STL triangle edges as dark opaque cylinders (mm radius). None = auto from mesh; 0 = off.
ILLUSTRATION_MESH_EDGE_RADIUS_MM = 0.01

# Illustration-only: scale catalog objects along their real rays (angle unchanged).
# camera_distance: optical (acquisition) pinhole–bear distance; 1 = catalog, >1 = farther.
# light_distance: catalog light–AABB-centre distance; 1 = catalog. Not POV key/fill lights.
ILLUSTRATION_CAMERA_DISTANCE = 0.5
ILLUSTRATION_LIGHT_DISTANCE = 1.2

# Ghost optical cameras on a world-z orbit (same |xy| as the illustrated pinhole).
# n = cameras on each side; step_deg = spacing. fade=True: transmit increases away from the present view.
ILLUSTRATION_ORBIT_CAMERAS = 17
ILLUSTRATION_ORBIT_STEP_DEG = 10.0
ILLUSTRATION_ORBIT_FADE = False

# Optical-camera frustum rays (four edges per camera): "all" | "single" | "none".
ILLUSTRATION_CAMERA_RAYS = "none"

# Per-view z-score anomaly image glued to the back of each optical camera (opposite the lens).
ILLUSTRATION_ANOMALY_PLATES = True
# Symmetric clip in σ for plate greyscale (smaller → more bear contrast). 2 matches a typical display clip.
ILLUSTRATION_ZSCORE_CLIP = 0.5
# Large copy of the z-score image from the optical camera whose plate appears biggest in this view.
# Shown as a stack of thin camera-body plates near the image centre (consecutive orbit views).
ILLUSTRATION_INSET_PLATE = True
ILLUSTRATION_INSET_STACK = 3

# PNG caption overlays (fractions of width/height; y from the top).
# Re-run this cell, then the last cell, to iterate without POV-Ray.
CAPTION_VIEWS_XY = (0.3, 0.96)
CAPTION_DEEP_LEARNING_XY = (0.53, 0.66)
CAPTION_LOCALIZATION_XY = (0.78, 0.88)

# Optional overrides (None = package defaults).
MANIFEST_PATH = None  # e.g. ROOT / "data" / "generated" / "m8_1" / "single_particle" / "<seq>" / "manifest.json"
DATA_ROOT = None  # default: data/generated/m8_1/single_particle

print(f"ROOT={display_path(ROOT)}")
print(f"SAMPLE_INDEX={SAMPLE_INDEX}  CAMERA_ANGLE_DEG={CAMERA_ANGLE_DEG}")
print(f"OUTPUT_POV={display_path(OUTPUT_POV)}  RENDER={RENDER}")
print(
    f"ILLUSTRATION_YAW_DEG={ILLUSTRATION_YAW_DEG}  "
    f"ILLUSTRATION_FOV_DEG={ILLUSTRATION_FOV_DEG}  "
    f"ILLUSTRATION_DISTANCE_SCALE={ILLUSTRATION_DISTANCE_SCALE}  "
    f"ILLUSTRATION_N_RAYS={ILLUSTRATION_N_RAYS}  "
    f"ILLUSTRATION_PARTICLE_RADIUS_MM={ILLUSTRATION_PARTICLE_RADIUS_MM}  "
    f"ILLUSTRATION_PARTICLE_LIGHT={ILLUSTRATION_PARTICLE_LIGHT}  "
    f"ILLUSTRATION_LIGHT_CONE_LENGTH_FRAC={ILLUSTRATION_LIGHT_CONE_LENGTH_FRAC}  "
    f"ILLUSTRATION_MESH_EDGE_RADIUS_MM={ILLUSTRATION_MESH_EDGE_RADIUS_MM}  "
    f"ILLUSTRATION_CAMERA_DISTANCE={ILLUSTRATION_CAMERA_DISTANCE}  "
    f"ILLUSTRATION_LIGHT_DISTANCE={ILLUSTRATION_LIGHT_DISTANCE}  "
    f"ILLUSTRATION_ORBIT_CAMERAS={ILLUSTRATION_ORBIT_CAMERAS}  "
    f"ILLUSTRATION_ORBIT_STEP_DEG={ILLUSTRATION_ORBIT_STEP_DEG}  "
    f"ILLUSTRATION_CAMERA_RAYS={ILLUSTRATION_CAMERA_RAYS}  "
    f"ILLUSTRATION_ANOMALY_PLATES={ILLUSTRATION_ANOMALY_PLATES}  "
    f"ILLUSTRATION_ZSCORE_CLIP={ILLUSTRATION_ZSCORE_CLIP}  "
    f"ILLUSTRATION_ORBIT_FADE={ILLUSTRATION_ORBIT_FADE}  "
    f"ILLUSTRATION_INSET_PLATE={ILLUSTRATION_INSET_PLATE}  "
    f"ILLUSTRATION_INSET_STACK={ILLUSTRATION_INSET_STACK}  "
    f"CAPTION_VIEWS_XY={CAPTION_VIEWS_XY}  "
    f"CAPTION_DEEP_LEARNING_XY={CAPTION_DEEP_LEARNING_XY}  "
    f"CAPTION_LOCALIZATION_XY={CAPTION_LOCALIZATION_XY}"
)
```

```python
result = export_m8_physical_scene(
    sample_index=SAMPLE_INDEX,
    camera_angle_deg=CAMERA_ANGLE_DEG,
    output_pov=OUTPUT_POV,
    manifest_path=MANIFEST_PATH,
    data_root=DATA_ROOT,
    repo_root_path=ROOT,
    render=RENDER,
    illustration_yaw_deg=ILLUSTRATION_YAW_DEG,
    illustration_fov_deg=ILLUSTRATION_FOV_DEG,
    illustration_distance_scale=ILLUSTRATION_DISTANCE_SCALE,
    illustration_n_rays=ILLUSTRATION_N_RAYS,
    illustration_particle_radius_mm=ILLUSTRATION_PARTICLE_RADIUS_MM,
    illustration_particle_light=ILLUSTRATION_PARTICLE_LIGHT,
    illustration_light_cone_length_frac=ILLUSTRATION_LIGHT_CONE_LENGTH_FRAC,
    illustration_mesh_edge_radius_mm=ILLUSTRATION_MESH_EDGE_RADIUS_MM,
    illustration_camera_distance=ILLUSTRATION_CAMERA_DISTANCE,
    illustration_light_distance=ILLUSTRATION_LIGHT_DISTANCE,
    illustration_orbit_cameras=ILLUSTRATION_ORBIT_CAMERAS,
    illustration_orbit_step_deg=ILLUSTRATION_ORBIT_STEP_DEG,
    illustration_camera_rays=ILLUSTRATION_CAMERA_RAYS,
    illustration_anomaly_plates=ILLUSTRATION_ANOMALY_PLATES,
    illustration_zscore_clip=ILLUSTRATION_ZSCORE_CLIP,
    illustration_orbit_fade=ILLUSTRATION_ORBIT_FADE,
    illustration_inset_plate=ILLUSTRATION_INSET_PLATE,
    illustration_inset_stack=ILLUSTRATION_INSET_STACK,
    illustration_caption_views_xy=CAPTION_VIEWS_XY,
    illustration_caption_deep_learning_xy=CAPTION_DEEP_LEARNING_XY,
    illustration_caption_localization_xy=CAPTION_LOCALIZATION_XY,
)
for key, path in result.items():
    print(f"{key}: {repo_relative_path(path)}")
print("If illumination rays are FALLBACK, see warnings above; add illustration_source_rays.json to use real segments.")
```

```python
# Caption-only: redraw labels on the last POV PNG (no ray-trace).
plain = ROOT / "outputs" / "renders" / "m8_physical_scene_plain.png"
png = ROOT / "outputs" / "renders" / "m8_physical_scene.png"
if not plain.is_file():
    raise FileNotFoundError(
        f"Missing {repo_relative_path(plain)}; run the previous cell with RENDER=True first."
    )
overlay_workflow_captions(
    png,
    source=plain,
    views_xy=CAPTION_VIEWS_XY,
    deep_learning_xy=CAPTION_DEEP_LEARNING_XY,
    localization_xy=CAPTION_LOCALIZATION_XY,
)
print(f"png: {repo_relative_path(png)}")
print(f"png_plain: {repo_relative_path(plain)}")
```
